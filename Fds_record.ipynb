Experiment 6: Time Series Analysis of Website Traffic

```python
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
from statsmodels.tsa.seasonal import seasonal_decompose

data = pd.read_csv('website_traffic.csv', parse_dates=['Date'], index_col='Date')
data['Page_Views'] = data['Page_Views'].interpolate(method='time')

decomposition = seasonal_decompose(
    data['Page_Views'], model='additive', period=7
)
fig = decomposition.plot()
fig.set_size_inches(10, 8)
plt.suptitle('Time Series Decomposition', y=1.02)
plt.show()

data['Page_Views_MA'] = data['Page_Views'].rolling(window=7).mean()

plt.figure(figsize=(10, 4))
plt.plot(data.index, data['Page_Views'], label='Page Views')
plt.plot(
    data.index,
    data['Page_Views_MA'],
    label='7-Day Moving Average',
    color='orange',
)
plt.title('Page Views with Moving Average')
plt.xlabel('Date')
plt.ylabel('Page Views')
plt.legend()
plt.grid(True)
plt.show()

data['Page_Views_Z'] = (
    data['Page_Views'] - data['Page_Views'].mean()
) / data['Page_Views'].std()
anomalies = data[data['Page_Views_Z'].abs() > 3]

plt.figure(figsize=(10, 4))
plt.plot(
    data.index, data['Page_Views'], label='Page Views', color='C0', alpha=0.8
)
plt.scatter(
    anomalies.index,
    anomalies['Page_Views'],
    color='red',
    label='Anomalies',
    zorder=5,
)
plt.title('Page Views Anomaly Detection')
plt.xlabel('Date')
plt.ylabel('Page Views')
plt.legend()
plt.grid(True)
plt.show()
```

Experiment 7:. Random Sampling and Sampling Distribution

```python
import matplotlib.pyplot as plt
import numpy as np
population_mean = 50
population_std = 10
population_size = 100000
population = np.random.normal(
    population_mean, population_std, population_size
)
sample_sizes = [30, 50, 100]
num_samples = 1000
sample_means = {}
for size in sample_sizes:
    sample_means[size] = []
    for _ in range(num_samples):
        sample = np.random.choice(population, size=size, replace=False)
        sample_means[size].append(np.mean(sample))
plt.figure(figsize=(12, 8))
for i, size in enumerate(sample_sizes):
    plt.subplot(len(sample_sizes), 1, i + 1)
    plt.hist(sample_means[size], bins=30, alpha=0.7, label=f"Sample Size {size}")
    plt.axvline(
        np.mean(population),
        color="red",
        linestyle="dashed",
        linewidth=1.5,
        label="Population Mean",
    )
    plt.title(f"Sampling Distribution (Sample Size {size})")
    plt.xlabel("Sample Mean")
    plt.ylabel("Frequency")
    plt.legend()
plt.tight_layout()
plt.show()
```

Lab Experiment 8: Conduct Z test for the Data Given using Python Objective

```python
import numpy as np
import scipy.stats as stats
sample_data = np.array([
    152, 148, 151, 149, 147, 153, 150, 148, 152, 149,
    151, 150, 149, 152, 151, 148, 150, 152, 149, 150,
    148, 153, 151, 150, 149, 152, 148, 151, 150, 153
])
population_mean = 150
sample_mean = np.mean(sample_data)
sample_std = np.std(sample_data, ddof=1)
n = len(sample_data)
z_statistic = (sample_mean - population_mean) / (sample_std / np.sqrt(n))
p_value = 2 * (1 - stats.norm.cdf(np.abs(z_statistic)))
print(f"Sample Mean: {sample_mean:.2f}")
print(f"Z-Statistic: {z_statistic:.4f}")
print(f"P-Value: {p_value:.4f}")
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: The average weight is significantly different from 150 grams.")
else:
    print("Fail to reject the null hypothesis: There is no significant difference in average weight from 150 grams.")
```

Exp No:12 Experiment to detect outliers in a given data set. Description: Understand the procedure to identify the outliers in a given dataset

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
%matplotlib inline
array = np.random.randint(1, 100, 16)
print("Original Array:", array)
print("Mean:", array.mean())
print("25th Percentile:", np.percentile(array, 25))
print("50th Percentile:", np.percentile(array, 50))
print("75th Percentile:", np.percentile(array, 75))
print("100th Percentile:", np.percentile(array, 100))
def outDetection(array):
    sorted(array)
    Q1, Q3 = np.percentile(array, [25, 75])
    IQR = Q3 - Q1
    lr = Q1 - (1.5 * IQR)
    ur = Q3 + (1.5 * IQR)
    return lr, ur
lr, ur = outDetection(array)
print("Lower Range:", lr, "Upper Range:", ur)
sns.displot(array)
sns.distplot(array)
plt.show()
new_array = array[(array > lr) & (array < ur)]
print("Array after 1st filtering:", new_array)
sns.displot(new_array)
plt.show()
lr1, ur1 = outDetection(new_array)
print("New Lower Range:", lr1, "New Upper Range:", ur1)
final_array = new_array[(new_array > lr1) & (new_array < ur1)]
print("Final Array:", final_array)
sns.distplot(final_array)
plt.show()
```

Exp No:14 Experiment to understand feature scaling

Exp No:18 Experiment to understand KNN algorithm for a given data set  

```python
import numpy as np
import pandas as pd
df = pd.read_csv('/content/Iris - Sheet1.csv')
df.info()
df.variety.value_counts()
df.head()
features = df.iloc[:, :-1].values
label = df.iloc[:, 4].values
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
xtrain, xtest, ytrain, ytest = train_test_split(features, label, test_size=0.2, random_state=0)
model_KNN = KNeighborsClassifier(n_neighbors=5)
model_KNN.fit(xtrain, ytrain)
print(model_KNN.score(xtrain, ytrain))
print(model_KNN.score(xtest, ytest))
from sklearn.metrics import confusion_matrix
confusion_matrix(label, model_KNN.predict(features))
from sklearn.metrics import classification_report
print(classification_report(label, model_KNN.predict(features)))
```

Exp No:19 Experiment to understand Logistic Regression for a given data set.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
import os
if not os.path.exists('Social_Network_Ads.csv'):
    np.random.seed(42)
    n_samples = 400
    mock_data = {
        'User ID': np.random.randint(15562261, 15815236, size=n_samples),
        'Gender': np.random.choice(['Male', 'Female'], size=n_samples),
        'Age': np.random.randint(18, 60, size=n_samples),
        'EstimatedSalary': np.random.randint(15000, 150000, size=n_samples),
        'Purchased': np.random.choice([0, 1], size=n_samples, p=[0.64, 0.36])
    }
    df = pd.DataFrame(mock_data)
    df.to_csv('Social_Network_Ads.csv', index=False)
    print("Local 'Social_Network_Ads.csv' generated successfully!")
df = pd.read_csv('Social_Network_Ads.csv')
df.head()
features = df.iloc[:, [2, 3]].values
label = df.iloc[:, 4].values
for i in range(1, 401):
    x_train, x_test, y_train, y_test = train_test_split(features, label, test_size=0.2, random_state=i)
    model = LogisticRegression()
    model.fit(x_train, y_train)
    train_score = model.score(x_train, y_train)
    test_score = model.score(x_test, y_test)
    if test_score > train_score:
        print("Test {} Train{} Random State {}".format(test_score, train_score, i))
x_train, x_test, y_train, y_test = train_test_split(features, label, test_size=0.2, random_state=26)
finalModel = LogisticRegression()
finalModel.fit(x_train, y_train)
print(finalModel.score(x_train, y_train))
print(finalModel.score(x_test, y_test))
print(classification_report(label, finalModel.predict(features)))
```

Exp No:20 Experiment to understand K-means clustering algorithm for a given data set.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
import os
%matplotlib inline
if not os.path.exists('Mall_Customers.csv'):
    np.random.seed(42)
    n_customers = 200
    mock_data = {
        'CustomerID': np.arange(1, n_customers + 1),
        'Genre': np.random.choice(['Male', 'Female'], size=n_customers),
        'Age': np.random.randint(18, 70, size=n_customers),
        'Annual Income (k$)': np.random.randint(15, 140, size=n_customers),
        'Spending Score (1-100)': np.random.randint(1, 100, size=n_customers)
    }
    df_mock = pd.DataFrame(mock_data)
    df_mock.to_csv('Mall_Customers.csv', index=False)
    print("Local 'Mall_Customers.csv' generated successfully!")
df = pd.read_csv('Mall_Customers.csv')
df.info()
df.head()
sns.pairplot(df)
plt.show()
features = df.iloc[:, [3, 4]].values
model = KMeans(n_clusters=5, random_state=42)
model.fit(features)
Final = df.iloc[:, [3, 4]].copy()
Final['label'] = model.predict(features)
Final.head()
sns.set_style("whitegrid")
sns.FacetGrid(Final, hue="label", height=8) \
   .map(plt.scatter, "Annual Income (k$)", "Spending Score (1-100)") \
   .add_legend()
plt.show()
features_el = df.iloc[:, [2, 3, 4]].values
wcss = []
for i in range(1, 10):
    model = KMeans(n_clusters=i, random_state=42)
    model.fit(features_el)
    wcss.append(model.inertia_)
plt.plot(range(1, 10), wcss)
plt.show()
```
